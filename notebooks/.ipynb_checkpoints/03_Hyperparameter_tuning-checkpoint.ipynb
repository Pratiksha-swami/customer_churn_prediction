```python
import pandas as pd 
import time
import joblib

from sklearn.model_selection import (train_test_split,GridSearchCV,RandomizedSearchCV,cross_val_score)
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

from sklearn.ensemble import(RandomForestClassifier,AdaBoostClassifier,GradientBoostingClassifier)
from xgboost import XGBClassifier
```

```python
df=pd.read_csv("../dataset/churn_clean.csv")

df.drop("customerID",axis=1,inplace=True)
df["Churn"]=df["Churn"].map({"Yes":1,"No":0})

df=pd.get_dummies(df,drop_first=True)
```

```python
X=df.drop("Churn",axis=1)
y=df["Churn"]

X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,stratify=y,test_size=0.20)
```

```python
scaler=StandardScaler()
numeric=["tenure","MonthlyCharges","TotalCharges"]

X_train[numeric]=scaler.fit_transform(X_train[numeric])
X_test[numeric]=scaler.transform(X_test[numeric])
```

```python
rf=RandomForestClassifier(random_state=41)

param_grid={
    "n_estimators":[100,200],
    "max_depth":[10,20,None],
    "min_samples_split":[2,5],
    "min_samples_leaf":[1,2]
}
```

```python
grid_rf=GridSearchCV(estimator=rf,param_grid=param_grid,cv=5,scoring="accuracy",n_jobs=-1)

grid_rf.fit(X_train,y_train)
```

```python
print("Best Parameter",grid_rf.best_params_)
print("Best cv score",grid_rf.best_score_)
```

```python
best_rf=grid_rf.best_estimator_

prediction=best_rf.predict(X_test)

print("Test Accuracy",accuracy_score(y_test,prediction))
```

```python
ada=AdaBoostClassifier(random_state=42)

param_grid={
    "n_estimators":[50,100,200],
    "learning_rate":[0.01,0.1,1]
}
```

```python
grid_ada=GridSearchCV(
    ada,
    param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_ada.fit(X_train,y_train)
```

```python
print(grid_ada.best_params_)
print(grid_ada.best_score_)
```

```python
best_ada=grid_ada.best_estimator_

prediction=best_ada.predict(X_test)

print("Test Accuracy:", accuracy_score(y_test,prediction))
```

```python
gb = GradientBoostingClassifier(random_state=42)

param_grid = {
    "n_estimators":[100,200],
    "learning_rate":[0.05,0.1],
    "max_depth":[3,5]
}
```

```python
grid_gb = GridSearchCV(
    gb,
    param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_gb.fit(X_train,y_train)
```

```python
best_gb = grid_gb.best_estimator_

prediction = best_gb.predict(X_test)

print("Test Accuracy:", accuracy_score(y_test,prediction))
```

```python
xgb = XGBClassifier(
    random_state=42,
    eval_metric="logloss"
)

param_dist = {
    "n_estimators":[100,200,300],
    "max_depth":[3,5,7],
    "learning_rate":[0.01,0.05,0.1],
    "subsample":[0.8,1.0],
    "colsample_bytree":[0.8,1.0]
}
```

```python
random_xgb = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist,
    n_iter=10,
    cv=5,
    scoring="accuracy",
    random_state=42,
    n_jobs=-1
)

random_xgb.fit(X_train,y_train)
```

```python
print(random_xgb.best_params_)
print(random_xgb.best_score_)
```

```python
best_xgb = random_xgb.best_estimator_

prediction = best_xgb.predict(X_test)

print("Test Accuracy:", accuracy_score(y_test,prediction))
```

```python
# lets compare all models now 

models={
    "Random Forest":best_rf,
    "AdaBoost":best_ada,
    "Gradient Boosting":best_gb,
    "XGBoost":best_xgb
}

cv_results=[]

for name,model in models.items():
    score=cross_val_score(model,X_train,y_train,cv=5,scoring="accuracy")
    cv_results.append({
        "Model":name,
        "CV Accuracy":score.mean()
    })
pd.DataFrame(cv_results).sort_values("CV Accuracy",ascending=False)
```

```python
joblib.dump(best_xgb,"../model/final_churn_model.pkl")
```
